```python
import os
import matplotlib.pyplot as plt
import torch
from pytomography import io
from pytomography.transforms.SPECT import SPECTAttenuationTransform, SPECTPSFTransform
from pytomography.algorithms import OSEM
from pytomography.projectors.SPECT import SPECTSystemMatrix
from pytomography.likelihoods import PoissonLogLikelihood
```

```python
# Paths are set in the data cell at the top of this tutorial
save_path = DATA / 'SPECT'
```

## Part 1: Opening Data

This tutorial involves reconstructing Lu-177 SPECT data from multiple bed positions and stitching the resulting reconstructions into a single reconstruction consisting of the full patient.

First we'll specify the projection data file paths: each one corresponds to a different bed position.

```python
files_NM = [
    os.path.join(save_path, 'Lu177-PSMA-GEDisc', 'bed1_projections.dcm'),
    os.path.join(save_path, 'Lu177-PSMA-GEDisc', 'bed2_projections.dcm'),
]
path_CT = os.path.join(save_path, 'Lu177-PSMA-GEDisc', 'CT')
files_CT = [os.path.join(path_CT, file) for file in os.listdir(path_CT)]
```

Lets print the energy window information so we can assign the correct indices to the photopeak and scatter

```python
_, _, proj_meta = io.SPECT.read_dicom(files_NM[0])
for i, window in enumerate(proj_meta.energy_windows):
    print(f'Index {i}:   Name: "{window.name}", Energies: {window.lower:g}-{window.upper:g} keV')
```

In this case since we are imaging the 208 keV peak, the index of the peak is 1, the lower scatter window's index is 3 and the upper one's 2, as confirmed by the `EnergyWindowInformationSequence`

Ideally, stitching of multiple bed positions would consist of (i) concatenating projections in projection space and (ii) reconstructing the concatenated projections. Unfortunately, the two bed positions may be acquired at different radial positions per projection, so the seperate bed positions need to be reconstructed seperately and then stitched together after reconstruction.

Note that if we load the projections seperately, with `io.SPECT.read_dicom` (it returns every energy window: `[window, angle, r, z]`)...

```python
projections_upper_fov, _, _ = io.SPECT.read_dicom(files_NM[0])
projections_lower_fov, _, _ = io.SPECT.read_dicom(files_NM[1])
```

...and then plot the photopeaks...

```python
# Plot [1,0] for index 1 (photopeak) and index 0 (angle 0)
plt.figure(figsize=(6,3))
plt.subplot(121)
plt.imshow(projections_lower_fov[1,0].cpu().T, vmax=40, cmap='magma', origin='lower')
plt.axis('off')
plt.subplot(122)
plt.imshow(projections_upper_fov[1,0].cpu().T, vmax=40, cmap='magma', origin='lower')
plt.axis('off')
plt.show()
```

...there are sharp cutoffs at the edges of the FOV. When reconstructing the data and then stitching together, this can lead to stitching artefacts at the edges of the field of view. 

Instead, we can take advantage of the fact that the projections are slightly overlapping to fill in missing data in each bed position. `io.SPECT.read_dicom_beds` reads every bed, sorts them from the most inferior to the most superior, and fills each bed's projections beyond the middle of the overlap with its neighbour's. It returns, for each bed, its projections, `object_meta` and `proj_meta`.

* *Note: at the edges, this means we may be mixing data from two different radial positions per projection (which means PSF modeling may be ever so slightly invalid). This, however, is still not nearly as egregious as having zeros in those locations. Since the stitching seams after reconstruction are at the central points of projection overlap of the projections, the end result of this has no noteworthy impact. (I'm certain the vendors do something similar to this, and if not, they really should.)*

```python
beds = io.SPECT.read_dicom_beds(files_NM)  # the inferior bed first
(projections_lower, object_meta_lower, proj_meta_lower), (projections_upper, object_meta_upper, proj_meta_upper) = beds
# Plot [1,0] for index 1 (photopeak) and index 0 (angle 0)
plt.figure(figsize=(6,3))
plt.subplot(121)
plt.imshow(projections_lower[1,0].cpu().T, vmax=40, cmap='magma', origin='lower')
plt.axis('off')
plt.subplot(122)
plt.imshow(projections_upper[1,0].cpu().T, vmax=40, cmap='magma', origin='lower')
plt.axis('off')
plt.show()
```

Before we reconstruct, lets examine the attenuation maps and corresponding projections to ensure the data is properly aligned

* Note the attenuation map for each set of projections is derived from a single set of CT files `files_CT` corresponding to a wholebody field of view. `io.attenuation_map` puts the CT on each bed's grid (`object_meta`), at the photon energy of the 208 keV window.

```python
attenuation_map_upper = io.attenuation_map(files_CT, object_meta_upper, energy=proj_meta_upper.energy_windows[1].photon_energy)
attenuation_map_lower = io.attenuation_map(files_CT, object_meta_lower, energy=proj_meta_lower.energy_windows[1].photon_energy)
plt.figure(figsize=(6,6))
plt.subplot(221)
plt.imshow(attenuation_map_upper.sum(axis=1).cpu().T, cmap='bone', origin='lower')
plt.axis('off')
plt.subplot(222)
plt.imshow(projections_upper[1,0].cpu().T, vmax=40, cmap='magma', origin='lower')
plt.axis('off')
plt.subplot(223)
plt.imshow(attenuation_map_lower.sum(axis=1).cpu().T, cmap='bone', origin='lower')
plt.axis('off')
plt.subplot(224)
plt.imshow(projections_lower[1,0].cpu().T, vmax=40, cmap='magma', origin='lower')
plt.axis('off')
plt.show()
```

## Part 2: Reconstruction

The following function performs reconstruction of a single bed position (if you are unfamiliar with this, please see the introductory DICOM tutorial). It takes the bed's projections, which `read_dicom_beds` has filled in from its neighbour, and the bed's own `object_meta` and `proj_meta`: `io.SPECT.energy_window_scatter` makes the scatter estimate from those projections, with the windows' widths from `proj_meta`. Before it reconstructs, it prints the memory the reconstruction will need (`estimate_memory`).


```python
def reconstruct_singlebed(projections, object_meta, proj_meta):
    # Change these depending on your file:
    index_peak = 1
    index_lower = 3
    index_upper = 2
    photopeak = projections[index_peak]
    scatter = io.SPECT.energy_window_scatter(projections, proj_meta, index_peak, index_lower, index_upper)
    # Build system matrix
    photon_energy = proj_meta.energy_windows[index_peak].photon_energy
    attenuation_map = io.attenuation_map(files_CT, object_meta, energy=photon_energy)
    psf_meta = io.SPECT.collimator_psf('GI-MEGP', photon_energy)
    att_transform = SPECTAttenuationTransform(attenuation_map)
    psf_transform = SPECTPSFTransform(psf_meta)
    # Create system matrix
    system_matrix = SPECTSystemMatrix(
        obj2obj_transforms = [att_transform,psf_transform],
        proj2proj_transforms= [],
        object_meta = object_meta,
        proj_meta = proj_meta)
    likelihood = PoissonLogLikelihood(system_matrix, photopeak, additive_term=scatter)
    reconstruction_algorithm = OSEM(likelihood)
    print(reconstruction_algorithm.estimate_memory(n_subsets=8))
    return reconstruction_algorithm(n_iters=4, n_subsets=8)
```

From this we can reconstruct each bed position seperately:

```python
recon_upper = reconstruct_singlebed(projections_upper, object_meta_upper, proj_meta_upper)
recon_lower = reconstruct_singlebed(projections_lower, object_meta_lower, proj_meta_lower)
```

We can view each reconstruction seperately

```python
plt.figure(figsize=(6,3))
plt.subplot(121)
plt.imshow(recon_upper.max(axis=1)[0].cpu().T, cmap='Greys', vmax=5, interpolation='gaussian', origin='lower')
plt.axis('off')
plt.subplot(122)
plt.imshow(recon_lower.max(axis=1)[0].cpu().T, cmap='Greys', vmax=5, interpolation='gaussian', origin='lower')
plt.axis('off')
plt.show()
```

To stitch these reconstructed images together, we use `io.SPECT.stitch_beds`. It takes the reconstructions with their beds' `object_meta`, in the same order, places each bed by its grid, and returns the stitched image with the `object_meta` of its own grid, which starts at the bottom slice of the lowest bed.

```python
recon_stitched, object_meta_stitched = io.SPECT.stitch_beds(
    recons=torch.stack([recon_upper, recon_lower]),
    object_metas=[object_meta_upper, object_meta_lower])
object_meta_stitched
```

Now we can view the stitched image

```python
maximum_intensity_projection = recon_stitched.max(axis=1)[0].cpu()
plt.figure(figsize=(3,6))
plt.imshow(maximum_intensity_projection.T, cmap='Greys', vmax=5, interpolation='gaussian', origin='lower')
plt.axis('off')
plt.show()
```

## Save the images

`io.save_dicom` writes the stitched image as a DICOM nuclear medicine (NM) file, in its place in the patient: `object_meta_stitched`, the grid `stitch_beds` returned. `reference=files_NM[1]` gives it the patient and study of the projections, so DICOM viewers show it with the CT.

```python
saved = io.save_dicom(recon_stitched, OUTPUT / 't_dicommultibed' / 'osem_stitched', object_meta_stitched,
                      reference=files_NM[1], series_description='OSEM 4x8, two beds stitched', overwrite=True)
```

```python

import json, platform, torch, pytomography
def _peak_ram_gb():
    try:
        import psutil
        m = psutil.Process().memory_info()
        if hasattr(m, "peak_pagefile"):  # Windows: peak committed memory
            return round(m.peak_pagefile / 1e9, 1)
    except ImportError:
        pass
    import resource  # ru_maxrss is in kB on Linux, bytes on macOS
    return round(resource.getrusage(resource.RUSAGE_SELF).ru_maxrss / (1e9 if platform.system() == "Darwin" else 1e6), 1)
_info = dict(pytomography=getattr(pytomography, "__version__", "?"), source=pytomography.__file__,
             torch=torch.__version__, python=platform.python_version(),
             gpu=torch.cuda.get_device_name(0) if torch.cuda.is_available() else "cpu",
             peak_gpu_gb=round(torch.cuda.max_memory_allocated() / 1e9, 2) if torch.cuda.is_available() else 0.0,
             peak_ram_gb=_peak_ram_gb())
print("__PYTOMOGRAPHY_RUN__" + json.dumps(_info))
```
