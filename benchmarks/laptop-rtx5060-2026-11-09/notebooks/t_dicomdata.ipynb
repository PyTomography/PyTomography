# DICOM Introduction

```python
from pytomography import datasets

# The tutorial data: downloaded the first time it runs (see Tutorial data in the docs)
datasets.fetch("SPECT/Lu177-NEMA-SymT2")
DATA = datasets.data_dir()  # the PYTOMOGRAPHY_DATA folder, or ~/pytomography_data
# Results go here, never into the data folder
OUTPUT = datasets.output_dir("SPECT/Lu177-NEMA-SymT2")
```

```python
import os
from pytomography import io
from pytomography.transforms.SPECT import SPECTAttenuationTransform, SPECTPSFTransform
from pytomography.algorithms import OSEM
from pytomography.projectors.SPECT import SPECTSystemMatrix
from pytomography.likelihoods import PoissonLogLikelihood
from pytomography.utils import print_collimator_parameters
import matplotlib.pyplot as plt
```

```python
# Paths are set in the data cell at the top of this tutorial
PATH = DATA / 'SPECT'
PATH = os.path.join(PATH, 'Lu177-NEMA-SymT2')
```

## Part 1: Opening Data

First we specify the required filepaths to open the SPECT scan (`file_NM`) and the CT data files (`files_CT`).

```python
# initialize the `path`` variable below to specify the location of the required data
path_CT = os.path.join(PATH, 'CT')
files_CT = [os.path.join(path_CT, file) for file in os.listdir(path_CT)]
file_NM = os.path.join(PATH, 'projection_data.dcm')
```

`io.SPECT.read_dicom` reads the projections and their metadata from the projection data file

* The metadata contains information such as the voxel size (mm), where the grid is in the patient, the number of angles, the scanner radius (mm) and every energy window.
* The grid is in the frame of the scan's CT. Siemens records its registration of the SPECT to the CT taken with it in the projection file (a rigid transform: here 4 mm and half a degree), and `read_dicom` moves the grid and turns the views by it, as Siemens does for its own reconstruction.

```python
projections, object_meta, proj_meta = io.SPECT.read_dicom(file_NM)
```

The projections hold every energy window of the file, `[window, angle, r, z]`. The windows are listed in `proj_meta.energy_windows`:

```python
for i, window in enumerate(proj_meta.energy_windows):
    print(f'Index {i}:   Name: "{window.name}", Energies: {window.lower:.1f}-{window.upper:.1f} keV')
```

In this case, since we are specifically imaging the 208 keV peak, we take index 0, the projections of the 187-228 keV energy window above.

```python
photopeak = projections[0]
photopeak.shape
```

The shape of the photopeak is $(N_{angles}, \text{Width}, \text{Height})$.

```python
plt.figure(figsize=(7,6))
plt.imshow(photopeak[0].cpu().T, cmap='nipy_spectral', origin='lower')
plt.axis('off')
plt.colorbar(label='Counts')
```

Scatter projections can be estimated using the triple energy window function `io.SPECT.energy_window_scatter`. One needs to specify the index of the peak, and lower/upper energy windows; their widths come from `proj_meta.energy_windows`

```python
scatter = io.SPECT.energy_window_scatter(projections, proj_meta, peak=0, lower=1, upper=2)
```

## Part 2: Modeling

### Attenuation Transform

The attenuation transform is used to "correct" for the fact that some photons get atteunated in the patient before they reach the detector. `SPECTAttenuationTransform` takes an attenuation map on the reconstruction grid, in units of mm $^{-1}$.

`io.attenuation_map` makes one from the CT: it reads the CT slices, converts their Hounsfield units to attenuation coefficients at the photon energy of the reconstructed window, and puts them on the grid of `object_meta`, each SPECT voxel the mean of the CT voxels inside it. Both images carry their place in the patient, so nothing has to be aligned by hand.

```python
attenuation_map = io.attenuation_map(files_CT, object_meta, energy=proj_meta.energy_windows[0].photon_energy)
att_transform = SPECTAttenuationTransform(attenuation_map)
```

The attenuation map is a tensor on the reconstruction grid, so it can be viewed directly.

Viewing the attenuation map allows for a consistency check to ensure that it's properly aligned with the projection data

```python
sample_slice = attenuation_map.cpu()[:,70].T
```

```python
plt.subplots(1,2,figsize=(7,3))
plt.subplot(121)
plt.title('Projection Data')
plt.imshow(photopeak[0].cpu().T, cmap='nipy_spectral', origin='lower')
plt.axis('off')
plt.colorbar(label='Counts')
plt.subplot(122)
plt.title('From CT Slices')
plt.imshow(sample_slice, cmap='Greys_r', origin='lower', interpolation='Gaussian')
plt.axis('off')
plt.colorbar()
```

### PSF Modeling

PSF modeling requires knowing the parameters of the collimators used for the scan. A list of all supported collimator types can be determined by uncommenting the following line and running it

```python
#print_collimator_parameters()
```

In our case, we know that this scan was performed on a Siemens Symbia scanner with medium energy collimators and 3/8'' crystals
* There is an option to use intrinsic resolution of the crystals of the scanner (if you know it); this has a minor impact on the reconstruction

```python
collimator_name = 'SY-ME'
photon_energy = proj_meta.energy_windows[0].photon_energy  # 208 keV
intrinsic_fwhm = 3.8  # mm
psf_meta = io.SPECT.collimator_psf(
    collimator_name,
    photon_energy,
    intrinsic_fwhm=intrinsic_fwhm
)
print(psf_meta)
```

Now we can build our PSF transform required to create the system matrix

```python
psf_transform = SPECTPSFTransform(psf_meta)
```

## System Matrix

Now that we have the required transforms, we can begin to build the system matrix. It requires:
1. `obj2obj_transforms`. These map from object space to object space. In the case of SPECT imaging, both the attenuation transform and PSF transform correspond to these types of transforms.
2. `proj2proj_transforms`: These map from projection to projection space. Though we have none here, they would be used to model PSF in PET imaging.
3. `object_meta` and `proj_meta`. These are used to initialize the transforms. For example, data from `object_meta` is used to align the attenuation map, and radial position from `proj_meta` is used to get proper blurring parameters for the PSF transform

```python
system_matrix = SPECTSystemMatrix(
        obj2obj_transforms = [att_transform,psf_transform],
        proj2proj_transforms = [],
        object_meta = object_meta,
        proj_meta = proj_meta)
```

The system matrix $H$ represents the imaging system and maps an object $f$ to a corresponding image $g$ under the imaging system

## Likelihood

```python
likelihood = PoissonLogLikelihood(system_matrix, photopeak, scatter)
```

## Reconstruct the object

We'll reconstruct the object using OSEM. For this, we need to specify the measured projection data (`photopeak`/`scatter`) and the `system_matrix`. Before it runs, `estimate_memory` says how much RAM and GPU memory it will need with 8 subsets:

```python
reconstruction_algorithm = OSEM(likelihood)
print(reconstruction_algorithm.estimate_memory(n_subsets=8))
```

We can reconstruct the object by calling the reconstruction algorithm for a given number of iterations/subsets.

```python
reconstructed_object = reconstruction_algorithm(n_iters=4, n_subsets=8)
```

We can compare this reconstruction to one done by the vendor. `io.load_image` reads the scanner's reconstruction with its own place in the patient, and `.on(object_meta)` puts it on our grid, so the two can be compared voxel by voxel. Both are in the frame of the CT: Siemens records its registration of the SPECT to the CT in the projection file, and `read_dicom` applies it to the grid, as Siemens does to its reconstruction.

* Note: we divide the vendor reconstruction by 96 because this particualr vendor multiplies by the number of acquisition angles when saving the data. (*some vendors alternatively divide by 48, the number of projection angles, so be careful*)

```python
recon_vendor = io.load_image(os.path.join(PATH, 'scanner_recon.dcm')).on(object_meta) / proj_meta.num_projections
```

We can looks at axial slices from all three reconstructions.

```python
idx_z = 61
slice_pytomography = reconstructed_object.cpu()[:,:,idx_z].T
slice_vendor = recon_vendor.cpu()[:,:,idx_z].T
```

```python
plt.subplots(1,2,figsize=(7,3), gridspec_kw={'wspace':0.0})
plt.subplot(121)
plt.imshow(slice_pytomography , cmap='magma', vmax=27, interpolation='gaussian', origin='lower')
plt.text(0.03, 0.97, 'PyTomography', color='white', fontsize=12, fontweight='bold', transform=plt.gca().transAxes, ha='left', va='top')
plt.xlim(30,98)
plt.ylim(98,40)
plt.axis('off')
plt.subplot(122)
plt.imshow(slice_vendor, cmap='magma', vmax=27, interpolation='gaussian', origin='lower')
plt.text(0.03, 0.97, 'Siemens Software', color='white', fontsize=12, fontweight='bold', transform=plt.gca().transAxes, ha='left', va='top')
plt.xlim(30,98)
plt.ylim(98,40)
plt.axis('off')
plt.show()
```

## Save the images

`io.save_dicom` writes the reconstruction as a DICOM nuclear medicine (NM) file, in its place in the patient (`object_meta.affine`). `reference=file_NM` gives it the patient and study of the projections, so DICOM viewers show it with the scan's CT.

**Note: some vendor reconstructions (i) multiply by the total number of projections and/or (ii) normalize to counts/second when they save their SPECT reconstructions. To compare with them, scale `reconstructed_object` the same way before you save it.**

```python
saved = io.save_dicom(reconstructed_object, OUTPUT / 't_dicomdata' / 'osem', object_meta, reference=file_NM,
                      series_description='OSEM 4x8', overwrite=True)
```
